```python
import pandas as pd
import seaborn as sns
import numpy as np
import matplotlib.pyplot as plt
```

```python
import os
import glob

path = os.path.join(os.getcwd(), 'data', 'raw', '*.csv')
path = glob.glob(path)
print(path)
```

```python
df = pd.read_csv(path[0])
df.head()
```

```python
duplicates = df.duplicated().sum()
print(f'Number of Duplicates: {duplicates}')
# No Duplicates
sns.heatmap(df.isnull(), cbar=False, cmap='viridis')
# No nulls
num_cols = df.select_dtypes(include=np.number).columns
for i in num_cols:
    sns.boxplot(x =df[i])
    plt.title(f'Boxplot {i}')
    plt.tight_layout()
    plt.show()
# There are outliers
# Will not remove them in this data
```
